```python
##25.   Diseña un agente para un juego de encender luces correctas en un panel.

##En este juego el agente debe encender las luces correctas en un panel de 3x3. El agente recibe una recompensa positiva por cada luz encendida que sea correcta y una recompensa negativa por cada luz encendida incorrectamente. Las luces incorrectas son las casillas que el agente debe evitar. El objetivo del agente es maximizar su recompensa total encendiendo las luces correctas y evitando las incorrectas.
```

```python
import numpy as np
import random


class Panel():
    def __init__(self):
        self.state = np.array([[1, -1, 1],
                               [-1, 1, -1],
                               [1, -1, 1]])

    def recompensa(self, row, col):
        return self.state[row, col]
    
    def valid_moves(self):
        return [(i, j) for i in range(3) for j in range(3)]


class Agent():
    def __init__(self, alpha=0.1, epsilon=0.1):
        #Q 
        self.Q = {} 
        self.alpha = alpha
        self.epsilon = epsilon

    def elegir_accion(self, panel):
        movimientos = panel.valid_moves()

        # exploración
        if np.random.uniform(0,1) < self.epsilon:
    
            return random.choice(movimientos)

        # explotación
        maxQ = -1000
        mejor_mov = movimientos[0]

        for mov in movimientos:
            valor = self.Q.get(mov, 0)
            if valor > maxQ:
                maxQ = valor
                mejor_mov = mov
        return mejor_mov

    def actualizar_Q(self, accion, recompensa):
        valor_actual = self.Q.get(accion, 0)
        self.Q[accion] = valor_actual + self.alpha * (recompensa - valor_actual)


class Game():
    def __init__(self, agent):
        self.agent = agent
        self.panel = Panel()
        self.recompensa_total = 0

    def jugar(self, turnos=100):

        for t in range(turnos):
            accion = self.agent.elegir_accion(self.panel)
            row, col = accion
            recompensa = self.panel.recompensa(row, col)
            self.agent.actualizar_Q(accion, recompensa)
            self.recompensa_total += recompensa



agente = Agent()
juego = Game(agente)

juego.jugar()
    
```
